```python
# ============================================================
# NODE 1 — PDF Classification + Extraction (FINAL)
# ============================================================

import re
import os
import pymupdf as fitz

DATA_DIR = "/content/drive/MyDrive/Colab Notebooks/enzymes"

TDS_SECTIONS = [
    "Product Description",
    "Effective material",
    "Activity",
    "Application",
    "Function",
    "Dosage",
    "Organoleptic",
    "Physicochemical",
    "Food Safety Data",
    "FOOD SAFTY DATA",
    "FOOD SAFETY DATA",
    "Package",
    "Storage",
]

SECTION_CANONICAL = {
    "FOOD SAFTY DATA":  "Food Safety Data",
    "FOOD SAFETY DATA": "Food Safety Data",
}

FOOTER_PATTERNS = [
    "VTR&beyond", "Pingbei", "Nanping", "Zhuhai",
    "Stresemann", "Berlin", "Tel:", "Mail:", "Website:",
    "Last updating", "info@vtrbeyond"
]

def is_footer(txt: str) -> bool:
    return any(pat in txt for pat in FOOTER_PATTERNS)


def detect_doc_type(doc, pdf_name: str) -> str:
    first_page_text = doc[0].get_text("text").upper()
    if "TECHNICAL DATA SHEET" in first_page_text:
        return "tds"
    if "TDS" in pdf_name.upper():
        return "tds"
    return "ref"


def extract_tds(doc, pdf_name: str) -> list:
    blocks = []

    for page in doc:
        page_blocks = page.get_text("dict")["blocks"]

        # ── Pass 1: find all headings with y position ────
        headings = []

        for b in page_blocks:
            if b["type"] != 0:
                continue
            for line in b["lines"]:
                line_txt  = ""
                line_size = 0
                line_bold = False
                line_y0   = line["bbox"][1]

                for span in line["spans"]:
                    txt = span["text"].strip()
                    if not txt:
                        continue
                    line_txt  += txt + " "
                    line_size  = max(line_size, round(span["size"], 1))
                    line_bold  = line_bold or bool(span["flags"] & 2**4)

                line_txt = line_txt.strip()
                if not line_txt:
                    continue

                if line_size >= 13.0 and line_bold:
                    continue

                is_main = line_size >= 12.0 and line_bold
                is_food = line_size >= 11.0 and line_bold and "FOOD SAF" in line_txt.upper()

                if is_main or is_food:
                    heading = line_txt.rstrip(":")
                    if "FOOD SAF" in heading.upper():
                        name = "Food Safety Data"
                    else:
                        name = SECTION_CANONICAL.get(heading, heading)
                    headings.append((line_y0, name))

        if not headings:
            continue

        # ── SORT headings by y position (fixes two-column PDFs) ──
        headings.sort(key=lambda x: x[0])

        # ── Pass 2: collect content by y-range ──────────
        for i, (h_y0, section_name) in enumerate(headings):
            next_y0       = headings[i + 1][0] if i + 1 < len(headings) else float("inf")
            content_parts = []

            for b in page_blocks:
                if b["type"] != 0:
                    continue
                for line in b["lines"]:
                    line_y0   = line["bbox"][1]
                    line_txt  = ""
                    line_size = 0
                    line_bold = False

                    for span in line["spans"]:
                        txt = span["text"].strip()
                        if not txt:
                            continue
                        line_txt  += txt + " "
                        line_size  = max(line_size, round(span["size"], 1))
                        line_bold  = line_bold or bool(span["flags"] & 2**4)

                    line_txt = line_txt.strip()
                    if not line_txt:
                        continue

                    if line_y0 <= h_y0:
                        continue
                    if line_y0 >= next_y0:
                        continue

                    if is_footer(line_txt):
                        continue

                    if line_size >= 13.0 and line_bold:
                        continue

                    if line_size >= 12.0 and line_bold:
                        continue

                    if line_size >= 11.0 and line_bold and "FOOD SAF" in line_txt.upper():
                        continue

                    if line_size >= 11.0 and line_bold:
                        content_parts.append(f"[{line_txt}]")
                        continue

                    if re.match(r"Last updating", line_txt, re.IGNORECASE):
                        continue

                    content_parts.append(line_txt)

            content = " ".join(content_parts).strip()
            content = re.sub(r"\bB\s*V\s*Zyme\b.*?(\s|$)", " ", content)
            content = re.sub(r"\b[A-Z]{2,}\d{3,}\b\s*®?", "", content)
            content = re.sub(r"®", "", content)
            content = re.sub(r"\s{2,}", " ", content).strip()

            if content:
                blocks.append({
                    "doc":      pdf_name,
                    "doc_type": "tds",
                    "section":  section_name,
                    "content":  content,
                    "raw":      content,
                    "page":     None,
                })

    return blocks


def extract_ref(doc, pdf_name: str) -> list:
    blocks          = []
    current_section = None
    current_content = []
    MIN_LENGTH      = 50

    def save_current():
        if current_section and current_content:
            merged = " ".join(current_content).strip()
            if len(merged) >= MIN_LENGTH:
                blocks.append({
                    "doc":      pdf_name,
                    "doc_type": "ref",
                    "section":  current_section,
                    "content":  merged,
                    "raw":      merged,
                    "page":     None,
                })

    for page in doc:
        for b in page.get_text("dict")["blocks"]:
            if b["type"] != 0:
                continue
            for line in b["lines"]:
                line_parts           = []
                line_is_main_heading = False
                line_is_sub_heading  = False
                line_is_metadata     = False
                heading_txt          = ""

                for span in line["spans"]:
                    txt     = span["text"].strip()
                    size    = span["size"]
                    is_bold = bool(span["flags"] & 2**4)

                    if not txt:
                        continue
                    if size <= 9.0:
                        line_is_metadata = True
                        break
                    if size >= 13.0 and is_bold:
                        line_is_main_heading = True
                        heading_txt          = txt
                        break
                    if size >= 11.5 and is_bold:
                        line_is_sub_heading = True
                        heading_txt         = txt
                        break
                    line_parts.append(txt)

                if line_is_metadata:
                    continue
                if line_is_main_heading:
                    save_current()
                    current_section = heading_txt
                    current_content = []
                elif line_is_sub_heading:
                    current_content.append(f"[{heading_txt}]")
                else:
                    if line_parts:
                        current_content.append(" ".join(line_parts))

    save_current()
    return blocks


# ============================================================
# MAIN — Run on all PDFs
# ============================================================

raw_blocks = []
failed     = []

all_pdfs = [f for f in os.listdir(DATA_DIR) if f.endswith(".pdf")]
print(f"PDFs found: {len(all_pdfs)}\n")

for pdf_name in all_pdfs:
    pdf_path = os.path.join(DATA_DIR, pdf_name)
    try:
        doc      = fitz.open(pdf_path)
        doc_type = detect_doc_type(doc, pdf_name)
        blocks   = extract_tds(doc, pdf_name) if doc_type == "tds" else extract_ref(doc, pdf_name)
        doc.close()
        raw_blocks.extend(blocks)
        print(f"[{doc_type.upper()}] {pdf_name} → {len(blocks)} blocks")
    except Exception as e:
        print(f"[ERROR] {pdf_name}: {e}")
        failed.append(pdf_name)

print(f"\n{'='*50}")
print(f"Total raw blocks : {len(raw_blocks)}")
print(f"  TDS : {sum(1 for b in raw_blocks if b['doc_type'] == 'tds')}")
print(f"  REF : {sum(1 for b in raw_blocks if b['doc_type'] == 'ref')}")
if failed:
    print(f"Failed : {failed}")
```

```python
# ============================================================
# NODE 2 — Cleaning + Normalization (FINAL BEST VERSION)
# ============================================================
# - General cleaning for both TDS + REF
# - TDS: deep field-aware normalization
# - REF: light clean, preserve French structure for Node 3
# - Bonus: Food Safety sub-section markers cleaned into
#          readable key:value pairs instead of [marker] noise
# - Input  : raw_blocks (from Node 1)
# - Output : cleaned_blocks
# ============================================================

import re

# ── General Cleaning ─────────────────────────────────────────

def clean_general(text: str) -> str:
    if not text:
        return ""
    # Unicode fixes
    text = text.replace("\u2013", "-").replace("\u2014", "-")
    text = text.replace("\u2019", "'").replace("\u2018", "'")
    text = text.replace("\xa0", " ").replace("\u200b", "")
    text = text.replace("：", ":").replace("，", ",")
    # Fix hyphenated word split across lines: "white- cream" → "white-cream"
    text = re.sub(r"-\s+(\S)", r"-\1", text)
    # Fix double numbered list artifacts: "1. 1." → "1."
    text = re.sub(r"(\d+)\.\s+\1\.", r"\1.", text)
    # Fix orphan list numbers: "6. 1." → "1."
    text = re.sub(r"^\d+\.\s+(\d+\.)", r"\1", text, flags=re.MULTILINE)
    # Collapse multiple spaces/tabs
    text = re.sub(r"[ \t]+", " ", text)
    # Collapse 3+ newlines
    text = re.sub(r"\n{3,}", "\n\n", text)
    # Strip each line
    text = "\n".join(line.strip() for line in text.splitlines())
    return text.strip()


# ── TDS: Sub-section marker cleanup ──────────────────────────

def clean_subsection_markers(text: str) -> str:
    """
    Converts noisy [SubHeading] markers into clean readable labels.
    [Microbiology] Total plate count... → Microbiology: Total plate count...
    [Heavy metals] Cadmium...           → Heavy metals: Cadmium...
    Keeps content readable and searchable.
    """
    # [Marker] followed by content → "Marker: content"
    text = re.sub(r"\[([^\]]+)\]\s*", r"\1: ", text)
    # Clean up double colons that may result: "Microbiology: : Total"
    text = re.sub(r":\s*:", ":", text)
    return text


# ── TDS: Value Normalization ──────────────────────────────────

def normalize_threshold(text: str) -> str:
    """
    Unifies threshold notations:
    'less than 5'  → '<5'
    '< 5'          → '<5'
    'not detected' → 'absent'
    'absent in 25g'→ 'absent'
    'ND'           → 'absent'
    """
    # "less than X unit" → "<X unit"
    text = re.sub(
        r"less\s+than\s+([\d,\.]+\s*\w*)",
        lambda m: "<" + m.group(1).strip(),
        text, flags=re.IGNORECASE
    )
    # "< X" → "<X" (remove space after operator)
    text = re.sub(r"<\s+([\d,\.]+)", r"<\1", text)
    text = re.sub(r">\s+([\d,\.]+)", r">\1", text)
    # "not detected" / "ND" / "absent in Xg" → "absent"
    text = re.sub(r"\bnot\s+detected\b", "absent", text, flags=re.IGNORECASE)
    text = re.sub(r"\bND\b", "absent", text)
    text = re.sub(r"absent\s+in\s+[\d]+\s*g", "absent", text, flags=re.IGNORECASE)
    return text


def normalize_range(text: str) -> str:
    """
    Unifies range notations:
    'from 5 to 30 ppm' → '5-30 ppm'
    '5 to 30 ppm'      → '5-30 ppm'
    '5 ~ 30'           → '5-30'
    """
    text = re.sub(
        r"from\s+([\d,\.]+)\s+to\s+([\d,\.]+)",
        r"\1-\2", text, flags=re.IGNORECASE
    )
    text = re.sub(
        r"([\d,\.]+)\s+to\s+([\d,\.]+)",
        r"\1-\2", text, flags=re.IGNORECASE
    )
    text = re.sub(r"([\d,\.]+)\s*~\s*([\d,\.]+)", r"\1-\2", text)
    return text


def normalize_duration(text: str) -> str:
    """
    '2 years' → '24 months'
    '1 year'  → '12 months'
    """
    def years_to_months(m):
        return f"{int(m.group(1)) * 12} months"
    text = re.sub(r"(\d+)\s+years?", years_to_months, text, flags=re.IGNORECASE)
    return text


def normalize_percentage(text: str) -> str:
    """
    '< 15 %' → '<15%'
    '15 %'   → '15%'
    '<15 %'  → '<15%'
    """
    text = re.sub(r"([\d,\.]+)\s+%", r"\1%", text)
    text = re.sub(r"<\s*([\d,\.]+)\s*%", r"<\1%", text)
    return text


def normalize_units(text: str) -> str:
    """
    Fixes spacing in large numbers and units:
    '50 000UFC' → '50000 UFC'
    '50 000'    → '50000'
    """
    # Remove space inside numbers: "50 000" → "50000"
    text = re.sub(r"(\d)\s(\d{3})\b", r"\1\2", text)
    # Ensure space between number and unit: "50000UFC" → "50000 UFC"
    text = re.sub(r"(\d)([A-Za-z]{2,})", r"\1 \2", text)
    return text


def normalize_temperature(text: str) -> str:
    """
    '< 20 °C' → '<20°C'
    '20 °C'   → '20°C'
    '( < 20°C)' → '(<20°C)'
    """
    text = re.sub(r"([\d,\.]+)\s*°\s*C", r"\1°C", text)
    text = re.sub(r"<\s*([\d,\.]+)°C", r"<\1°C", text)
    return text


def normalize_tds(text: str, section: str) -> str:
    """
    Apply all TDS normalizations in order.
    Section-aware: skip certain normalizations for sections
    where they could cause false positives.
    """
    # Clean sub-section markers first
    text = clean_subsection_markers(text)
    # Apply all value normalizations
    text = normalize_threshold(text)
    text = normalize_range(text)
    text = normalize_percentage(text)
    text = normalize_temperature(text)
    text = normalize_units(text)
    # Duration only in Storage/Package sections
    if section in ("Storage", "Package"):
        text = normalize_duration(text)
    return text


# ── REF: Light Cleaning ───────────────────────────────────────

def clean_ref(text: str) -> str:
    """
    Light cleaning for REF docs.
    Keeps French text intact for Node 3 translation.
    Only cleans sub-section markers and whitespace.
    """
    # Convert [SubSection] markers to readable format
    text = re.sub(r"\[([^\]]+)\]\s*", r"\1: ", text)
    text = re.sub(r":\s*:", ":", text)
    return text


# ── Per-block Cleaning ───────────────────────────────────────

def clean_block(block: dict) -> dict:
    cleaned = block.copy()
    content = clean_general(block["content"])

    if block["doc_type"] == "tds":
        content = normalize_tds(content, block.get("section", ""))
    else:
        content = clean_ref(content)

    cleaned["content"] = content
    return cleaned


# ── MAIN ─────────────────────────────────────────────────────

cleaned_blocks = [clean_block(b) for b in raw_blocks]

# ── Verify ───────────────────────────────────────────────────

print(f"Total cleaned blocks : {len(cleaned_blocks)}")
print(f"  TDS : {sum(1 for b in cleaned_blocks if b['doc_type'] == 'tds')}")
print(f"  REF : {sum(1 for b in cleaned_blocks if b['doc_type'] == 'ref')}")

print("\n── TDS: Food Safety Data (marker cleanup) ──")
for b in cleaned_blocks:
    if b["doc_type"] == "tds" and b["section"] == "Food Safety Data":
        print(f"  {b['doc']}")
        print(f"  {b['content'][:400]}")
        break

print("\n── TDS: Storage (normalization) ──")
for b in cleaned_blocks:
    if b["doc_type"] == "tds" and b["section"] == "Storage":
        print(f"  {b['doc']} → {b['content']}")
        break

print("\n── TDS: Dosage ──")
for b in cleaned_blocks:
    if b["doc_type"] == "tds" and b["section"] == "Dosage":
        print(f"  {b['doc']} → {b['content']}")
        break

print("\n── REF: sub-section markers ──")
for b in cleaned_blocks:
    if b["doc_type"] == "ref":
        print(f"  {b['section']} → {b['content'][:300]}")
        break
```

```python
# ============================================================
# NODE 3 — Translation (REF French → English)
# ============================================================
# - REF blocks  → translate content to English via Google
# - TDS blocks  → skip, copy content directly to content_en
# - Input  : cleaned_blocks (from Node 2)
# - Output : translated_blocks (new field: content_en)
# ============================================================

!pip install deep-translator -q

from deep_translator import GoogleTranslator
import time

translator = GoogleTranslator(source="fr", target="en")

# Google Translate has a 5000 char limit per request
MAX_CHARS = 4500

def translate_text(text: str) -> str:
    """
    Translates French text to English.
    Splits into chunks if text exceeds Google's character limit.
    """
    if not text or len(text.strip()) == 0:
        return text

    # If short enough, translate directly
    if len(text) <= MAX_CHARS:
        try:
            return translator.translate(text)
        except Exception as e:
            print(f"  [translation error]: {e}")
            return text  # fallback: return original

    # Split into chunks at sentence boundaries
    sentences   = re.split(r'(?<=[.!?])\s+', text)
    chunks      = []
    current     = ""

    for sentence in sentences:
        if len(current) + len(sentence) <= MAX_CHARS:
            current += " " + sentence
        else:
            if current:
                chunks.append(current.strip())
            current = sentence

    if current:
        chunks.append(current.strip())

    # Translate each chunk and join
    translated_parts = []
    for chunk in chunks:
        try:
            translated_parts.append(translator.translate(chunk))
            time.sleep(0.3)  # avoid rate limiting
        except Exception as e:
            print(f"  [chunk translation error]: {e}")
            translated_parts.append(chunk)  # fallback

    return " ".join(translated_parts)


# ── MAIN ─────────────────────────────────────────────────────

translated_blocks = []

ref_count = sum(1 for b in cleaned_blocks if b["doc_type"] == "ref")
print(f"Blocks to translate : {ref_count} REF")
print(f"Blocks to skip      : {sum(1 for b in cleaned_blocks if b['doc_type'] == 'tds')} TDS\n")

for i, block in enumerate(cleaned_blocks):
    b = block.copy()

    if block["doc_type"] == "ref":
        print(f"  Translating [{block['section']}] ({len(block['content'])} chars)...")
        b["content_en"] = translate_text(block["content"])
        time.sleep(0.5)  # be gentle with the API
    else:
        # TDS already English
        b["content_en"] = block["content"]

    translated_blocks.append(b)

# ── Verify ───────────────────────────────────────────────────

print(f"\nTotal translated blocks : {len(translated_blocks)}")
print(f"  TDS : {sum(1 for b in translated_blocks if b['doc_type'] == 'tds')}")
print(f"  REF : {sum(1 for b in translated_blocks if b['doc_type'] == 'ref')}")

print("\n── REF Translation Samples ──")
for b in translated_blocks:
    if b["doc_type"] == "ref":
        print(f"\n  Section  : {b['section']}")
        print(f"  Original : {b['content'][:150]}")
        print(f"  English  : {b['content_en'][:150]}")
        print("  " + "─"*50)
```

```python
# ============================================================
# NODE 4 — Chunking
# ============================================================
# - TDS  → one chunk per section, enriched with metadata
# - REF  → one chunk per section, enriched with metadata
# - Input  : translated_blocks (from Node 3)
# - Output : chunks (list of dicts, ready for embedding)
# ============================================================

import uuid

def make_chunk_id(doc: str, section: str, index: int) -> str:
    base = f"{doc}_{section}_{index}"
    return str(uuid.uuid5(uuid.NAMESPACE_DNS, base))

chunks = []

for i, block in enumerate(translated_blocks):
    chunk = {
        "id":         make_chunk_id(block["doc"], block.get("section", ""), i),
        "doc":        block["doc"],
        "doc_type":   block["doc_type"],
        "section":    block.get("section", ""),
        "content":    block["content"],        # original (French if REF)
        "content_en": block["content_en"],     # English (used for embedding)
        "raw":        block.get("raw", ""),
    }
    chunks.append(chunk)

# ── Verify ───────────────────────────────────────────────────

print(f"Total chunks : {len(chunks)}")
print(f"  TDS : {sum(1 for c in chunks if c['doc_type'] == 'tds')}")
print(f"  REF : {sum(1 for c in chunks if c['doc_type'] == 'ref')}")

print("\n── Sample TDS chunk ──")
for c in chunks:
    if c["doc_type"] == "tds":
        print(f"  id       : {c['id']}")
        print(f"  doc      : {c['doc']}")
        print(f"  section  : {c['section']}")
        print(f"  content  : {c['content_en'][:150]}")
        break

print("\n── Sample REF chunk ──")
for c in chunks:
    if c["doc_type"] == "ref":
        print(f"  id       : {c['id']}")
        print(f"  doc      : {c['doc']}")
        print(f"  section  : {c['section']}")
        print(f"  content  : {c['content_en'][:150]}")
        break
```

```python
# ============================================================
# NODE 5 — Embedding + Index
# ============================================================
# - Embeds all chunks using all-MiniLM-L6-v2
# - Uses content_en (English only) for all embeddings
# - Stores chunks + embeddings in an index
# - Input  : chunks (from Node 4)
# - Output : index dict {chunks, embeddings}
# ============================================================

!pip install sentence-transformers -q

import pickle
import numpy as np
from sentence_transformers import SentenceTransformer

MODEL_NAME = "all-MiniLM-L6-v2"
EMBED_DIM  = 384
INDEX_PATH = "enzyme_index.pkl"

print(f"Loading model: {MODEL_NAME}")
embedder = SentenceTransformer(MODEL_NAME)

# Extract English texts for embedding
texts = [c["content_en"] for c in chunks]

print(f"Embedding {len(texts)} chunks...")
embeddings = embedder.encode(
    texts,
    batch_size=64,
    show_progress_bar=True,
    normalize_embeddings=True  # required for cosine similarity
)

print(f"Embeddings shape : {embeddings.shape}")
assert embeddings.shape == (len(chunks), EMBED_DIM), "Dimension mismatch!"

# Build index
index = {
    "chunks":     chunks,
    "embeddings": embeddings,
}

# Save to disk
with open(INDEX_PATH, "wb") as f:
    pickle.dump(index, f)
print(f"Index saved → {INDEX_PATH}")

# ── Verify ───────────────────────────────────────────────────

print(f"\nTotal indexed chunks : {len(index['chunks'])}")
print(f"Embedding matrix     : {index['embeddings'].shape}")
print(f"Sample norm          : {np.linalg.norm(embeddings[0]):.4f}")  # should be ~1.0

print("\n── Sample embeddings ──")
for i, c in enumerate(chunks[:3]):
    print(f"  [{i}] {c['doc']} | [{c['section']}] | norm={np.linalg.norm(embeddings[i]):.4f}")
```

```python
# ============================================================
# NODE 5.5 — Auto Synonym Expansion from Corpus
# ============================================================

!pip install -q keybert sentence-transformers

from keybert import KeyBERT
from sentence_transformers import SentenceTransformer
import numpy as np
from collections import defaultdict
import re

kw_model = KeyBERT(model=embedder)

# ── Step 1: Extract keywords from every chunk ─────────────
print("Extracting keywords from corpus...")

chunk_keywords = {}
for i, chunk in enumerate(chunks):
    text = chunk["content_en"]
    if not text or len(text) < 20:
        continue
    keywords = kw_model.extract_keywords(
        text,
        keyphrase_ngram_range=(1, 2),
        stop_words="english",
        top_n=5
    )
    chunk_keywords[i] = [kw for kw, score in keywords if score > 0.3]

print(f"Keywords extracted from {len(chunk_keywords)} chunks")


# ── Step 2: Find semantically similar keyword pairs ────────
print("\nFinding synonym pairs...")

all_keywords = list(set(
    kw for kws in chunk_keywords.values() for kw in kws
))

kw_embeddings = embedder.encode(
    all_keywords,
    normalize_embeddings=True,
    show_progress_bar=True
)

similarity_matrix = kw_embeddings @ kw_embeddings.T
SIMILARITY_THRESHOLD = 0.75

synonym_pairs = []
for i in range(len(all_keywords)):
    for j in range(i+1, len(all_keywords)):
        sim = similarity_matrix[i][j]
        kw_i = all_keywords[i]
        kw_j = all_keywords[j]
        if sim >= SIMILARITY_THRESHOLD and kw_i != kw_j:
            synonym_pairs.append((kw_i, kw_j, float(sim)))

synonym_pairs.sort(key=lambda x: x[2], reverse=True)

print(f"\nFound {len(synonym_pairs)} synonym pairs")
print("\n── Top 20 pairs ──")
for a, b, score in synonym_pairs[:20]:
    print(f"  {score:.3f} | '{a}' ↔ '{b}'")


# ── Step 3: Build expansion dictionary ────────────────────
print("\nBuilding expansion dictionary...")

EXPANSION_DICT = defaultdict(set)

for kw_a, kw_b, score in synonym_pairs:
    EXPANSION_DICT[kw_a].add(kw_b)
    EXPANSION_DICT[kw_b].add(kw_a)

# Manual French → English bridge terms
FRENCH_BRIDGE = {
    "douceur":       {"softness", "crumb"},
    "conservation":  {"storage", "shelf life"},
    "quantité":      {"dosage", "ppm"},
    "fermentation":  {"gassing power", "proofing"},
    "extensibilité": {"extensibility", "gluten"},
    "volume":        {"loaf volume", "oven spring"},
    "texture":       {"crumb texture", "elasticity"},
    "congélation":   {"frozen dough", "stability"},
    "tolérance":     {"tolerance", "stability"},
    "farine":        {"flour", "wheat"},
    "pain":          {"bread", "bakery"},
    "pâte":          {"dough"},
    "fraîcheur":     {"freshness", "softness"},
    "allergène":     {"allergen", "gluten"},
    "stockage":      {"storage", "temperature"},
    "humidité":      {"moisture"},
    "sécurité":      {"food safety"},
    "couleur":       {"color", "aspect"},
}

for fr, en_set in FRENCH_BRIDGE.items():
    EXPANSION_DICT[fr].update(en_set)

print(f"Total expansion terms: {len(EXPANSION_DICT)}")


# ── Step 4: Expand query function (language-aware) ────────
def expand_query(query: str, lang: str = "en") -> str:
    """
    Expands a query with corpus-derived synonyms.
    Only applies to non-English queries to avoid
    diluting English embeddings.
    """
    if lang == "en":
        return query  # English queries stay clean

    expanded = query
    query_lower = query.lower()

    for term, synonyms in EXPANSION_DICT.items():
        if term in query_lower:
            expanded += " " + " ".join(synonyms)

    return expanded.strip()


# ── Verify ────────────────────────────────────────────────
print("\n── Expansion Test ──")
test_queries = [
    ("Which enzyme improves gluten extensibility?", "en"),
    ("Dosage recommandé pour xylanase",             "fr"),
    ("bread softness improvement",                  "en"),
    ("fermentation tolerance enzyme",               "en"),
    ("Xylanase améliore extensibilité pâte",        "fr"),
]

for q, lang in test_queries:
    expanded = expand_query(q, lang)
    print(f"\n  Lang     : {lang}")
    print(f"  Original : {q}")
    print(f"  Expanded : {expanded}")
```

```python
# ============================================================
# NODE 6 — Query Router (FINAL BEST VERSION)
# ============================================================
# Strategy: Hybrid rule + classifier
#
# Layer 1 (FAST): Regex check for known product codes
#   → product code found in query → Path 1 instantly
#   → no model needed, 100% reliable
#
# Layer 2 (SMART): NLP classifier for functional queries
#   → no product code → classify intent
#   → "searching for products that match a function" → Path 2
#   → confidence < threshold → ask to rephrase
#
# This makes the classifier only handle cases it's good at
# (distinguishing functional queries from gibberish)
# instead of trying to detect product names (regex is better)
#
# Input  : query (string)
# Output : route ("path1" | "path2" | "rephrase")
# ============================================================

from transformers import pipeline
import re

CLASSIFIER_MODEL     = "cross-encoder/nli-MiniLM2-L6-H768"
CONFIDENCE_THRESHOLD = 0.50

print(f"Loading classifier: {CLASSIFIER_MODEL}")
classifier = pipeline(
    "zero-shot-classification",
    model=CLASSIFIER_MODEL
)

CANDIDATE_LABELS = [
    "searching for products that match a function",  # → Path 2
    "random or unclear input",                       # → rephrase
]

# ── Layer 1: Build product registry from index ───────────────

def build_product_registry(chunks: list) -> dict:
    product_codes = set()
    for c in chunks:
        doc = c["doc"].upper()
        # Standard codes: "HCB708", "SOFT405"
        codes = re.findall(r"[A-Z]{2,}\d{3,}", doc)
        for code in codes:
            product_codes.add(code)
        # Spaced codes: "GOX 110" → "GOX110", "TG 883" → "TG883"
        spaced = re.findall(r"([A-Z]{2,})\s+(\d{3,})", doc)
        for letters, digits in spaced:
            product_codes.add(letters + digits)

    aliases = {
        "ASCORBICACID": "acide ascorbique.pdf",
        "ASCORBIQUE":   "acide ascorbique.pdf",
        "E300":         "acide ascorbique.pdf",
        "VITAMINC":     "acide ascorbique.pdf",
        "VITAMINEC":    "acide ascorbique.pdf",
        "ASCORBIC":     "acide ascorbique.pdf",
    }
    return {"codes": product_codes, "aliases": aliases}

# Build registry once from index
registry = build_product_registry(index["chunks"])
print(f"Product registry: {len(registry['codes'])} codes + {len(registry['aliases'])} aliases")
print(f"Codes: {sorted(registry['codes'])}")


def detect_product_in_query(query: str, registry: dict):
    """
    Layer 1: Fast regex product detection.
    Returns product reference (code or doc name) or None.
    """
    query_clean = query.upper().replace(" ", "")

    # Check aliases first (multi-word product names)
    for alias, doc in registry["aliases"].items():
        if alias in query_clean:
            return doc

    # Check product codes
    for code in registry["codes"]:
        if code in query_clean:
            return code

    return None


# ── Layer 2: NLP classifier for non-product queries ──────────

def classify_functional_query(query: str) -> dict:
    """
    Layer 2: NLP classifier.
    Only called when no product code found.
    Distinguishes functional queries from gibberish/unclear input.
    """
    result    = classifier(query, candidate_labels=CANDIDATE_LABELS)
    top_label = result["labels"][0]
    top_score = result["scores"][0]

    # If top score is below threshold → genuinely unclear
    if top_score < CONFIDENCE_THRESHOLD:
        return {"route": "rephrase", "confidence": top_score, "label": top_label}

    # If classified as functional → Path 2
    if top_label == CANDIDATE_LABELS[0]:
        return {"route": "path2", "confidence": top_score, "label": top_label}

    # If classified as random/unclear → rephrase
    return {"route": "rephrase", "confidence": top_score, "label": top_label}


# ── Main Router ───────────────────────────────────────────────

def route_query(query: str) -> dict:
    """
    Hybrid router:
    1. Check for product code/name → Path 1 (no classifier needed)
    2. No product found → NLP classifier → Path 2 or rephrase
    """
    # ── Layer 1: product detection ───────────────────────────
    product_ref = detect_product_in_query(query, registry)

    if product_ref:
        return {
            "route":      "path1",
            "label":      "product detected",
            "confidence": 1.0,        # regex is deterministic
            "product":    product_ref,
            "query":      query
        }

    # ── Layer 2: NLP classifier ──────────────────────────────
    classification = classify_functional_query(query)

    return {
        "route":      classification["route"],
        "label":      classification["label"],
        "confidence": classification["confidence"],
        "product":    None,
        "query":      query
    }


# ── Verify ───────────────────────────────────────────────────

test_queries = [
    # Should be Path 1 (product detected)
    "What is the dosage of BVZyme HCB708?",
    "Is HCB710 GMO free?",
    "What is the storage condition for AMG880?",
    "Tell me about ascorbic acid dosage",
    "What is the function of SOFT405?",
    "Give me all info on GOX110",
    # Should be Path 2 (functional)
    "Which enzyme improves gluten extensibility?",
    "What enzymes can be used for bread volume improvement?",
    "Which enzyme is good for frozen dough?",
    "Which product helps with bread softness?",
    "What improves fermentation tolerance?",
    # Should be rephrase (unclear)
    "xyz blah blah random",
    "hello",
    "?????",
]

print("\n── Routing Test ──")
for q in test_queries:
    result = route_query(q)
    product_str = f" [{result['product']}]" if result.get("product") else ""
    print(f"\n  Query      : {q}")
    print(f"  Route      : {result['route']}{product_str}")
    print(f"  Confidence : {result['confidence']:.3f}")
```

```python
# ============================================================
# NODE 7 — Path 1: Structured Product Lookup (FINAL)
# ============================================================

import re

def extract_product_code(query: str, chunks: list):
    query_clean = query.upper().replace(" ", "")

    # Aliases for REF docs
    ALIASES = {
        "ASCORBICACID": "acide ascorbique.pdf",
        "ASCORBIQUE":   "acide ascorbique.pdf",
        "E300":         "acide ascorbique.pdf",
        "VITAMINC":     "acide ascorbique.pdf",
    }
    for alias, doc in ALIASES.items():
        if alias in query_clean:
            return doc

    # Extract product codes from all doc names
    product_codes = set()
    for c in chunks:
        doc   = c["doc"].upper()
        codes = re.findall(r"[A-Z]{2,}\d{3,}", doc)
        for code in codes:
            product_codes.add(code)

    for code in product_codes:
        if code in query_clean:
            return code

    return None


# Section keywords for TDS docs
SECTION_KEYWORDS_TDS = {
    "dosage":      "Dosage",
    "dose":        "Dosage",
    "ppm":         "Dosage",
    "storage":     "Storage",
    "store":       "Storage",
    "keep":        "Storage",
    "shelf":       "Storage",
    "allergen":    "Food Safety Data",
    "gmo":         "Food Safety Data",
    "safety":      "Food Safety Data",
    "microb":      "Food Safety Data",
    "heavy metal": "Food Safety Data",
    "moisture":    "Physicochemical",
    "activity":    "Activity",
    "function":    "Function",
    "application": "Application",
    "description": "Product Description",
    "material":    "Effective material",
    "ingredient":  "Effective material",
    "package":     "Package",
    "packaging":   "Package",
    "color":       "Organoleptic",
    "colour":      "Organoleptic",
    "aspect":      "Organoleptic",
}

# Section keywords for REF docs (French section names)
SECTION_KEYWORDS_REF = {
    "dosage":       "Dosages Recommandés (ppm*)",
    "dose":         "Dosages Recommandés (ppm*)",
    "ppm":          "Dosages Recommandés (ppm*)",
    "conversion":   "Table de Conversion Rapide (en grammes)",
    "storage":      "Recommandations pour ta Production",
    "store":        "Recommandations pour ta Production",
    "regulation":   "Réglementation",
    "legal":        "Réglementation",
    "authorized":   "Réglementation",
    "alternative":  "Alternatives et Complémentarité",
    "advantage":    "Avantages et Limitations",
    "limitation":   "Avantages et Limitations",
    "specification":"Spécifications Techniques",
    "property":     "Résumé Général",
    "properties":   "Résumé Général",
    "summary":      "Résumé Général",
    "function":     "Résumé Général",
    "reference":    "Références",
}


def path1_lookup(query: str, index: dict) -> list:
    chunks      = index["chunks"]
    product_ref = extract_product_code(query, chunks)

    if not product_ref:
        return []

    # Match chunks by doc name
    if product_ref.endswith(".pdf"):
        matched = [c for c in chunks if c["doc"] == product_ref]
    else:
        matched = [
            c for c in chunks
            if product_ref in c["doc"].upper().replace(" ", "")
        ]

    if not matched:
        return []

    # Pick keyword map based on doc type
    is_ref      = matched[0]["doc_type"] == "ref"
    keyword_map = SECTION_KEYWORDS_REF if is_ref else SECTION_KEYWORDS_TDS
    query_lower = query.lower()

    target_section = None
    for keyword, section in keyword_map.items():
        if keyword in query_lower:
            target_section = section
            break

    if target_section:
        section_match = [c for c in matched if c["section"] == target_section]
        if section_match:
            return section_match

    # No specific section → return all
    return matched


# ── Verify ───────────────────────────────────────────────────

print("── Path 1 Lookup Test ──")
test_queries_p1 = [
    "What is the dosage of HCB708?",
    "Is HCB710 GMO free?",
    "What is the storage condition for AMG880?",
    "Tell me about ascorbic acid dosage",
    "What is the function of SOFT405?",
    "What are the properties of ascorbic acid?",
]

for q in test_queries_p1:
    results = path1_lookup(q, index)
    print(f"\n  Query   : {q}")
    print(f"  Matched : {len(results)} chunk(s)")
    for r in results:
        print(f"    [{r['section']}] {r['content_en'][:120]}")
```

```python
# ============================================================
# NODE 8 — Path 2: Semantic Search
# ============================================================
# - Triggered when route == "path2"
# - Embeds query using all-MiniLM-L6-v2
# - Cosine similarity against all chunk embeddings
# - Returns Top-K=3 most similar chunks
# - Input  : query, index
# - Output : top 3 chunks with similarity scores
# ============================================================

import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

TOP_K = 3

def path2_semantic_search(query: str, index: dict, top_k: int = TOP_K) -> list:
    chunks     = index["chunks"]
    embeddings = index["embeddings"]

    # Embed query with same normalization as index
    query_embedding = embedder.encode(
        [query],
        normalize_embeddings=True
    )

    # Cosine similarity: (1, 384) x (N, 384).T → (1, N)
    scores = cosine_similarity(query_embedding, embeddings)[0]

    # Get top-k indices sorted by score descending
    top_indices = np.argsort(scores)[::-1][:top_k]

    results = []
    for idx in top_indices:
        results.append({
            "chunk": chunks[idx],
            "score": float(scores[idx]),
        })

    return results


# ── Verify ───────────────────────────────────────────────────

print("── Path 2 Semantic Search Test ──")
test_queries_p2 = [
    "Which enzyme improves gluten extensibility?",
    "What improves bread volume?",
    "enzyme for frozen dough",
    "What enzymes can be used for bread softness?",
    "Which product helps with fermentation tolerance?",
]

for q in test_queries_p2:
    results = path2_semantic_search(q, index)
    print(f"\n  Query : {q}")
    for i, r in enumerate(results):
        print(f"  [{i+1}] score={r['score']:.4f} | {r['chunk']['doc']} | [{r['chunk']['section']}]")
        print(f"       {r['chunk']['content_en'][:120]}")
```

```python
# ============================================================
# NODE 9 — Full Pipeline (Router + Retrieval) + Multilingual
# ============================================================

from typing import Dict, Any
from deep_translator import GoogleTranslator

try:
    from deep_translator import single_detection
except Exception:
    single_detection = None


def _detect_lang(text: str) -> str:
    if single_detection is None:
        return "en"
    try:
        lang = single_detection(text)
        return lang if isinstance(lang, str) and len(lang) >= 2 else "en"
    except Exception:
        return "en"


def _to_english(text: str, src_lang: str) -> str:
    if not text or src_lang == "en":
        return text
    try:
        return GoogleTranslator(source="auto", target="en").translate(text)
    except Exception:
        return text


def _from_english(text: str, tgt_lang: str) -> str:
    if not text or tgt_lang == "en":
        return text
    try:
        return GoogleTranslator(source="en", target=tgt_lang).translate(text)
    except Exception:
        return text


def retrieve(query: str) -> Dict[str, Any]:
    # ── Step 0: Detect + translate query ────────────────────
    query_lang = _detect_lang(query)
    query_en   = _to_english(query, query_lang)

    # ── Step 0.5: Expand query with corpus synonyms ─────────
    query_en = expand_query(query_en)

    # ── Step 1: Route ────────────────────────────────────────
    routing = route_query(query_en)

    print(f"\n{'='*60}")
    print(f"Query (orig): {query}")
    print(f"Lang       : {query_lang}")
    print(f"Query (en) : {query_en}")
    print(f"Route      : {routing['route']} (confidence={routing['confidence']:.3f})")

    # ── Step 2: Rephrase ─────────────────────────────────────
    if routing["route"] == "rephrase":
        msg_en = "Your question is ambiguous. Please rephrase it and try again."
        msg    = _from_english(msg_en, query_lang)
        print("  ⚠ Query is ambiguous. Please rephrase and try again.")
        return {
            "route":    "rephrase",
            "lang":     query_lang,
            "query":    query,
            "query_en": query_en,
            "message":  msg,
            "results":  []
        }

    # ── Step 3: Path 1 (exact lookup) ───────────────────────
    if routing["route"] == "path1":
        results = path1_lookup(query, index)

        if not results:
            print("  ⚠ Product not found in index, falling back to Path 2...")
            raw        = path2_semantic_search(query_en, index)
            results    = raw
            route_used = "path2_fallback"
        else:
            results    = [{"chunk": r, "score": 1.0} for r in results]
            route_used = "path1"

    # ── Step 4: Path 2 (semantic search) ────────────────────
    else:
        results    = path2_semantic_search(query_en, index)
        route_used = "path2"

    # ── Step 5: Translate results back ───────────────────────
    for r in results:
        chunk      = r.get("chunk", {})
        content_en = chunk.get("content_en", "")
        chunk["content"] = _from_english(content_en, query_lang)

    # ── Step 6: Print results ────────────────────────────────
    print(f"Route used : {route_used}")
    print(f"Results    : {len(results)} chunk(s)")

    for i, r in enumerate(results):
        chunk = r["chunk"]
        score = r["score"]
        print(f"\n  [{i+1}] score={score:.4f}")
        print(f"       doc     : {chunk.get('doc')}")
        print(f"       section : {chunk.get('section')}")
        print(f"       content : {chunk.get('content','')[:200]}")

    return {
        "route":    route_used,
        "lang":     query_lang,
        "query":    query,
        "query_en": query_en,
        "results":  results
    }
```

```python
queries = [
    ("What is the recommended dosage?",          "en"),
    ("Dosage recommandé pour les enzymes",        "fr"),
    ("Alpha-amylase utilisation ppm farine",      "fr"),
    ("Xylanase améliore extensibilité pâte",      "fr"),
    ("recommended dosage flour enzymes",          "en"),
]

for q, lang in queries:
    print(f"\n{'='*60}")
    print(f"Query: {q}")
    routing = route_query(q)
    print(f"Route: {routing['route']} (confidence={routing['confidence']:.3f})")

    if routing["route"] == "path1":
        results = path1_lookup(q, index)
        results = [{"chunk": r, "score": 1.0} for r in results]
    elif routing["route"] == "path2":
        q_expanded = expand_query(_to_english(q, lang), lang)  # ← expansion here
        results = path2_semantic_search(q_expanded, index)
    else:
        print("⚠ Rephrase needed")
        continue

    for i, r in enumerate(results):
        chunk = r["chunk"]
        score = r["score"]
        print(f"\n  [{i+1}] score={score:.4f}")
        print(f"       doc     : {chunk['doc']}")
        print(f"       section : {chunk['section']}")
        print(f"       content : {chunk['content_en'][:150]}")
```

```python
# =========================================
# NODE  — RECHERCHE SÉMANTIQUE (PostgreSQL)
# =========================================

!pip install -q sentence-transformers psycopg2-binary pgvector

import psycopg2
import numpy as np
from sentence_transformers import SentenceTransformer

# ← Fill in when credentials are available
DB_CONFIG = None

# ─────────────────────────────────────────────────────────────

embedder = SentenceTransformer("all-MiniLM-L6-v2")

def recherche_semantique(question: str, top_k: int = 3) -> list:
    if DB_CONFIG is None:
        print("⚠ Base de données non configurée. Veuillez remplir DB_CONFIG.")
        return []

    q_vec = embedder.encode(
        [question], convert_to_numpy=True, normalize_embeddings=True
    ).astype(np.float32)[0]

    vecteur_str = "[" + ",".join(f"{v:.8f}" for v in q_vec) + "]"

    conn = psycopg2.connect(**DB_CONFIG)
    with conn.cursor() as cur:
        cur.execute("""
            SELECT texte_fragment,
                   1 - (vecteur <=> %s::vector) AS score
            FROM embeddings
            ORDER BY vecteur <=> %s::vector
            LIMIT %s
        """, (vecteur_str, vecteur_str, top_k))
        rows = cur.fetchall()
    conn.close()

    return [{"rang": i+1, "texte": row[0], "score": float(row[1])}
            for i, row in enumerate(rows)]


def afficher_resultats(question, resultats):
    if not resultats:
        print("\n⚠ Aucun résultat — base de données non configurée.")
        return

    print("\n" + "═" * 65)
    print(f"  Question : {question}")
    print("═" * 65)
    for r in resultats:
        print(f"\n  Résultat {r['rang']}")
        print(f"  Score   : {r['score']:.4f}")
        print(f"  Texte   : {r['texte'][:400]}")
        print("  " + "─" * 62)
    print("═" * 65)


# ── Test ──────────────────────────────────────────────────────
question = "Améliorant de panification : quelles sont les quantités recommandées d'alpha-amylase, xylanase et d'Acide ascorbique ?"
resultats = recherche_semantique(question)
afficher_resultats(question, resultats)
```
